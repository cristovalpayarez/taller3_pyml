# 👁️ ¿Qué es la visión artificial?

La **visión artificial** es una tecnología que permite a las computadoras **“ver” y analizar imágenes o videos** para obtener información, de manera similar a como lo hacen nuestros ojos y cerebro. 🤖📷

## 🔍 ¿Para qué sirve?

La visión artificial puede utilizarse para:

* 📷 **Reconocer objetos** en fotografías.
* 🙂 **Identificar rostros**.
* 🚗 **Detectar vehículos y peatones**.
* 🏭 **Detectar defectos** en productos de una fábrica.
* 🩻 **Analizar imágenes médicas** para ayudar a los profesionales.
* 📱 **Reconocer textos** dentro de imágenes.

## 🧠 ¿Cómo funciona?

La visión artificial utiliza técnicas de **inteligencia artificial (IA)** y **aprendizaje automático** para aprender a reconocer patrones en imágenes y videos.

### 🐱🐶 Ejemplo

Si mostramos a un sistema miles de imágenes de **gatos y perros**, la computadora puede aprender sus características. Después, al recibir una imagen nueva, puede intentar determinar si contiene un gato 🐱 o un perro 🐶.

## 💡 En resumen

> **La visión artificial permite que las máquinas interpreten y comprendan información visual del mundo.** 🌎🤖


# 👤 Detección de rostros con Python usando OpenCV

La **detección de rostros** es una aplicación de la **visión artificial** que permite identificar automáticamente rostros humanos dentro de una imagen o video. 🤖📷

## 🐍 ¿Qué es OpenCV?

**OpenCV (Open Source Computer Vision Library)** es una biblioteca muy utilizada para trabajar con imágenes y videos.

Con OpenCV podemos realizar diferentes tareas, como:

* 👤 Detectar rostros.
* 👀 Detectar ojos.
* 📷 Capturar imágenes desde una cámara.
* 🎥 Procesar videos.
* 🔎 Analizar imágenes.
* 🖼️ Aplicar diferentes efectos y transformaciones.

## ⚙️ ¿Cómo funciona la detección de rostros?

Un método clásico de OpenCV utiliza **clasificadores Haar Cascade**. Estos clasificadores permiten analizar una imagen y buscar patrones que coincidan con las características de un rostro.

El proceso básico es:

1. 📷 **Capturar** una imagen o video mediante una cámara.
2. ⚫ **Convertir** la imagen a escala de grises.
3. 🔍 **Analizar** la imagen utilizando un clasificador.
4. 👤 **Detectar** las zonas donde se encuentran los rostros.
5. 🟩 **Marcar** los rostros detectados con un rectángulo.

## 🧠 Conceptos importantes

### 🔹 Clasificador Haar Cascade

Es un modelo utilizado por OpenCV para detectar diferentes objetos y características en imágenes. En este caso, se puede utilizar para localizar **rostros humanos**.

### 🔹 Escala de grises

Las imágenes se pueden convertir a tonos de gris antes de analizarlas. Esto simplifica la información que debe procesar el sistema y puede facilitar la detección.

### 🔹 Detección en tiempo real

Cuando se utiliza una cámara, OpenCV puede analizar los cuadros del video continuamente y detectar rostros mientras la cámara está funcionando. 🎥👤

## 👤 Detección vs. reconocimiento facial

Es importante diferenciar estos dos conceptos:

* 🔍 **Detección facial:** determina **dónde hay un rostro** en una imagen.
* 🧑 **Reconocimiento facial:** intenta determinar **a quién pertenece** ese rostro.

Por ejemplo, si una cámara encuentra tres rostros, la detección facial puede marcar los tres con rectángulos, pero eso no significa que sepa quiénes son las personas.

```python
# 1. Intalacion de Opencv Library

#pip install opencv-python


#imortacion de la libreria 
import cv2

# cargar la imagen 
imgpath = 'cierrese-encima-retrato-mujer-hermosa-maquillaje-natural-que-presenta-camara-tocar-su-cara-aislada-sobre-blanco_171337-659_1.jpg'
#  leer la imagen
img = cv2.imread(imgpath)

# mostrar la imagen
img.shape

print("Version:", cv2.__version__)
print("CascadeClassifier:", hasattr(cv2, "CascadeClassifier"))
```

```python
# convertimos la imagen a escala de grises
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow('Imagen en escala de grises', gray_img)
```

```python
# cargar el clasificador de rostros 

cascade = "haarcascade_frontalface_default.xml"
face_cascade = cv2.CascadeClassifier(cascade)

#verificar si el clasificador cargo correctamente

if face_cascade.empty():
    print("Error: No se pudo cargar.")
    
else:
    print ("Clasificador cargado correctamente")
```

```python
# 9. Detectar rostros en la imagen
face = face_cascade.detectMultiScale(gray_img, scaleFactor=1.1, minNeighbors=5, minSize=(50, 50))
```

```python
# 10 recorrer todos los rostros detectados y dibujar un rectangulo alrededor de ellos

for (x, y, w, h) in face:
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 2)

# 11. La imagenes se cargan en BGR, se debe convertir a RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
# Importar matplotlib
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 10))
plt.imshow(img_rgb)
plt.axis('off')
```

Captura de video en tiempo real

```python
# 1. Acceder a la camara de nuestro pc y leer un flujo de datos en tiemo real 


video_capture = cv2.VideoCapture(0)

# Funcion para detectar rostro con flujo de video y dibujar un recuadro delimitador 

def detect_bounding_box(vid):
    gray_img = cv2.cvtColor(vid, cv2.COLOR_BGR2GRAY)
    face = face_cascade.detectMultiScale(gray_img, 1.1, 5, minSize=(50, 50))
    for (x, y, w, h) in face:
        cv2.rectangle(vid, (x, y), (x + w, y + h), (0, 255, 0), 2)
    return face
```

```python
# 3. Buce while para detectar caras en real time

while True:
    result, video_frame = video_capture.read()
    if result is False:
        break
    faces = detect_bounding_box(video_frame)

    cv2.imshow("Mi cara detectada", video_frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

video_capture.release()
cv2.destroyAllWindows()
```
